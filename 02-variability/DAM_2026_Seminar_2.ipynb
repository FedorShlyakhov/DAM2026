If we have several conditions we can use & (and), | (or)

People who are younger than 18 (kids) or older than 60 (old people)

```python
df[(df["Age"]<18) |  (df["Age"]>60)]
```

Women who are adults:

```python
df[(df['Sex'] == 'female') & (df['Age'] >= 18)]
```

We can combine & and | and use `()`

Women who were in 1 or 2 class:

```python
df[((df['Pclass'] == 1) | (df['Pclass'] == 2)) & (df['Sex']=='female')]
```

Something more complex - women from 18 to 25 who embarked in Cherbourg(С), or Southampton (S)

```python
df[((df['Embarked'] =='C') | (df['Embarked'] =='S')) & (df['Sex'] == 'female') &
     (df['Age'] >= 18) & (df['Age'] <= 25)]
```

**Task** Create column with BMI of an athlete

$BMI = \frac{m}{h^2}$

where $m$ - weight in kg, $h$ - height in meters.

```python
#your code here
```

Standard cetegories for BMI:

*   $< 18.5$ - Underweight
*   $[18.5; 25.0)$  - Normal
*   $[25.0; 30.0)$ - Overweight
*   $>= 30.0$ - Obese


Compute the BMI category for every athelete, save to `Category_BMI` column (if BMI was NaN - category should also be NaN). Compute number of athletes in every category.

```python
#your code here
def bmi_categ(bmi):
  if bmi < 18.5:
    return 'Underweight'
  if 18.5 <= bmi < 25:
    return 'Normal'
  if 25 <= bmi < 30:
    return 'Overweight'
  if bmi >= 30:
    return  'Obese'

  return np.nan

data['Category_BMI'] = data['BMI'].apply(bmi_categ)
data.sample(4)
```

```python
data['Category_BMI'].value_counts()
```

```python
data[(data['Category_BMI'] == 'Obese') & (data['Medal'] == 'Gold')]
```

```python
data[(data['Category_BMI'] == 'Obese') & (data['Medal'] == 'Gold')].groupby('Sport')['Name'].nunique().sort_values(ascending=False)
```
